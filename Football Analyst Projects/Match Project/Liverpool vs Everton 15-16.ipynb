# Formations Plot

```python
# Variable To Show Specific Match 
events = sb.events(match_id = 3754294)['tactics']

# Shows Df Starting With First Instance
events[0]
```

```python
# Looks Through Statsbomb's Open Data
parser = Sbopen()

# Retrieve Detailed Event Data And Unpacks It To Four Variables:
event, related, freeze, tactics = parser.event(3754294)

# Starting Players For Liverpool
starting_xi_event = event.loc[((event['type_name'] == 'Starting XI') &
                               (event['team_name'] == 'Liverpool')), ['id', 'tactics_formation']]

# Merge 'starting_xi_event' On 'id' In 'tatics' Column 
starting_xi = tactics.merge(starting_xi_event, on='id')

# Creating Shortened Names For Players From Data
player_short_names = {'Simon Mignolet ': 'Simon Mignolet ',
                      'Alberto Moreno Pérez': 'Alberto Moreno',
                      'Mamadou Sakho': 'Mamadou Sakho',
                      'Dejan Lovren': 'Dejan Lovren',
                      'Nathaniel Edwin Clyne': 'Nathaniel Clyne',
                      'Lucas Pezzini Leiva': 'Lucas Leiva',
                      'James Philip Milner': 'James Milner',
                      'Philippe Coutinho Correia': 'Philippe Coutinho',
                      'Roberto Firmino Barbosa de Oliveira': 'Roberto Firmino',
                      'Adam David Lallana': 'Adam Lallana',
                      'Divock Okoth Origi': 'Divock Origi'}

# Variabele To Replace Name With Shortened Name
starting_xi['player_name'] = starting_xi['player_name'].replace(player_short_names)

# Add Players That Filter Only Successfull Ball Receipts From The Starting XI 
event = event.loc[((event['type_name'] == 'Ball Receipt') &
                   (event['outcome_name'].isnull()) &
                   (event['player_id'].isin(starting_xi['player_id']))
                   ), ['player_id', 'x', 'y']]

# Merge Events Onto The Starting Positions
event = event.merge(starting_xi, on='player_id')

# Print Positions
formation = event['tactics_formation'].iloc[0]
```

```python
# Styling The Pitch, Choosing Rotation And Size
pitch = VerticalPitch(pitch_color='grass', goal_type='line',line_color='white')

# Style Pitch And Text For Markers
fig, ax = pitch.draw(figsize=(6, 8.72))

# Style Pitch And Text For Markers
ax_text = pitch.formation(formation, positions=starting_xi.position_id, kind='text',
                          text=starting_xi.player_name.str.replace(' ', '\n'),
                          va='center', ha='center', fontsize=16, ax=ax)

# Styling Markers With Size, Location And Color
ax_scatter = pitch.formation(formation, positions=starting_xi.position_id, kind='scatter',
                             c='#ff0000', linewidth=3, s=500,
                             # Provide A Single Offset Instead Of A List
                             # For *xoffset* And *yoffset*
                             xoffset=7,
                             ax=ax)
```
